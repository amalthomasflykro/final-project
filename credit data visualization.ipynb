<h2 style="color:#1e3a8a;">1. Introduction</h2>

<p style="text-align:justify; line-height:1.8; color:#1e293b;">
Credit risk refers to the possibility that a borrower may fail to repay a loan or meet their financial obligations. Financial institutions use credit risk analysis to evaluate the financial condition, repayment capacity, credit history and borrowing behaviour of customers before approving loans.
</p>

<p style="text-align:justify; line-height:1.8; color:#1e293b;">
Credit risk analysis plays an important role in banking and financial services by helping institutions identify potential risks, make informed lending decisions and reduce financial losses.
</p>

<p style="text-align:justify; line-height:1.8; color:#1e293b;">
Analysing customer information such as credit scores, annual income, current loan amounts and monthly debt enables financial institutions to assess the creditworthiness of borrowers.
</p>

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

Univariate analysis – Loan Term

```python
plt.figure(figsize=(10, 5))

sns.histplot(data=df5, x='Current_Loan_Amount', bins=30, kde=True)

plt.title('Distribution of Current Loan Amount')
plt.xlabel('Current Loan Amount')
plt.ylabel('Frequency')

plt.show()
```

Observation:

The distribution of loan amounts indicates how customer borrowing is concentrated across different loan ranges. The shape of the distribution reveals whether most customers borrow relatively smaller amounts or whether loan amounts are spread across a wider range.

Key Findings:
- Identify the loan amount range with the highest customer concentration.
- Explain whether the distribution is positively or negatively skewed.
- Highlight any unusual patterns or extreme values.
- Explain what the distribution suggests about customer borrowing behaviour.

```python
# Calculate customer count for each loan term
loan_distribution = df5['Term'].value_counts()

# Create area chart
plt.figure(figsize=(10, 6))

plt.fill_between(
    loan_distribution.index,
    loan_distribution.values,
    alpha=0.35,
    color='#3498DB'
)

plt.plot(
    loan_distribution.index,
    loan_distribution.values,
    marker='o',
    linewidth=2.5,
    markersize=8,
    color='#2471A3'
)

plt.title(
    'Distribution of Customers by Loan Term',
    fontsize=16,
    fontweight='bold',
    pad=18
)

plt.xlabel('Loan Term', fontsize=12)
plt.ylabel('Number of Customers', fontsize=12)

plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

Observation

The chart compares the number of customers belonging to each loan-term category. It provides an overview of customer preferences between short-term and long-term borrowing.

Key Findings


The loan category with the highest customer count was identified based on the distribution of loan terms.



The difference in customer count and percentage between Short Term and Long Term loans was analysed.



The dominant borrowing pattern was determined by comparing the proportion of customers across loan terms.

```python
plt.figure(figsize=(10, 5))

sns.histplot(data=df5, x='Credit Score', bins=30, kde=True, color='teal')

plt.title('Distribution of Customer Credit Scores')
plt.xlabel('Credit Score')
plt.ylabel('Number of Customers')

plt.show()
```

Observation


The histogram illustrates the distribution of credit scores among customers. It helps identify the concentration of customers across different credit-score ranges and the overall pattern of their credit profiles.

Key Findings


Determine the credit-score range containing the highest number of customers.



Compare the distribution of customers across different credit-score ranges.



Identify unusually low and high credit-score values.



Assess the overall credit-score pattern to understand the credit profiles of customers.

```python
income_groups = pd.cut(
    df5['Annual Income'],
    bins=[0, 25000, 50000, 75000, 100000, float('inf')],
    labels=['Below 25K', '25K–50K', '50K–75K', '75K–100K', 'Above 100K']
)

income_distribution = income_groups.value_counts().sort_index()

plt.figure(figsize=(8, 8))

plt.pie(
    income_distribution,
    labels=income_distribution.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'width': 0.4, 'edgecolor': 'white'}
)

plt.title('Annual Income Distribution')
plt.tight_layout()
plt.show()
```

Observation

The  illustrates the distribution of annual income among borrowers. It helps identify income concentration, variations and the overall distribution pattern.

Key Findings

- The annual income range containing the largest number of customers.
- Whether customer income is concentrated in lower or higher ranges.
- Whether the distribution is positively or negatively skewed.
- Any unusually high or low income values.

```python
# Divide Monthly Debt into ranges
debt_groups = pd.cut(
    df5['Monthly Debt'],
    bins=5,
    labels=['Very Low', 'Low', 'Medium', 'High', 'Very High']
)

# Create a pivot table
debt_heatmap = pd.crosstab(
    debt_groups,
    df5['Home_Ownership']
)

# Plot heatmap
plt.figure(figsize=(10, 6))

sns.heatmap(
    debt_heatmap,
    annot=True,
    fmt='d',
    cmap='Blues',
    linewidths=0.5,
    linecolor='white'
)

plt.title(
    'Customer Monthly Debt Distribution by Home Ownership',
    fontsize=15,
    fontweight='bold',
    pad=15
)

plt.xlabel('Home Ownership', fontsize=12)
plt.ylabel('Monthly Debt Range', fontsize=12)

plt.tight_layout()
plt.show()
```

Observation

The histogram illustrates the distribution of monthly debt among customers. It helps examine how financial obligations vary across the customer base and identify the concentration of debt values.

Key Findings
- The monthly debt range with the highest frequency.
- Whether most customers have relatively low or high monthly debt.
- The degree of variation in customers' monthly financial obligations.
- Whether the distribution shows significant skewness.

```python
# Group credit history into year ranges
credit_history_groups = (
    df5['Years of Credit History']
    .round()
    .value_counts()
    .sort_index()
)

# Create line chart
plt.figure(figsize=(12, 6))

sns.lineplot(
    x=credit_history_groups.index,
    y=credit_history_groups.values,
    marker='o',
    linewidth=2.5,
    markersize=6,
    color='#2878B5'
)

plt.title(
    'Distribution of Customers by Years of Credit History',
    fontsize=16,
    fontweight='bold',
    pad=18
)

plt.xlabel('Years of Credit History', fontsize=12)
plt.ylabel('Number of Customers', fontsize=12)

plt.grid(True, linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

Observation


The histogram represents the distribution of customers based on the length of their credit history. It highlights differences in credit experience across the customer base.

Key Findings

- The most common credit-history duration.
- Whether customers generally have shorter or longer credit histories.
- The variation in credit experience.
- Any unusual values in the distribution.
